```python
# ใช้ฟอนต์ไทย (เลือกอย่างใดอย่างหนึ่ง)
plt.rcParams['font.family'] = 'Tahoma'   # ใช้ง่ายสุดใน Windows
plt.rcParams['axes.unicode_minus'] = False
```

```python
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='MAJOR', order=df['MAJOR'].value_counts().index)
plt.title('Distribution of Major')
plt.xlabel('Major')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.countplot(data=df, y='ENT_METHOD_DESC', order=df['ENT_METHOD_DESC'].value_counts().index)
plt.title('Distribution of Entry Method')
plt.xlabel('Entry Method')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.countplot(data=df, y='NATIONALITY', order=df['NATIONALITY'].value_counts().index, palette='husl')
plt.title("Types of Nationalities Students Have")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='TCAS', order=df['TCAS'].value_counts().index, palette='Set2')
plt.title('Distribution of TCAS')
plt.xlabel('TCAS')
plt.ylabel('Count')
plt.show()
```

```python
pivot_major_year = pd.crosstab(df['MAJOR'], df['ADMIT_YEAR'])
plt.figure(figsize=(10,6))
sns.heatmap(pivot_major_year, annot=True, fmt='d')
plt.title('จำนวนนักศึกษาแต่ละสาขา แยกตามปีที่รับเข้า')
plt.xlabel('ปีที่รับเข้า')
plt.ylabel('สาขา')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.countplot(data=df, y='STATUS_DESC_THAI', order=df['STATUS_DESC_THAI'].value_counts().index, palette='Set3')
plt.title('Distribution of Status')
plt.xlabel('Status')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='ADMIT_YEAR', hue='TITLE_NAME_ENG', multiple='stack')
plt.title('Distribution of Admission Year by Title')
plt.show()
```
